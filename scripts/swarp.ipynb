```python
Hphi=np.loadtxt('Hphi.txt')
Hphi.shape
```

```python
phi=Hphi[:,0]
H=Hphi[:,1]
iphi0=np.where(phi==0)
inds=iphi0[0]
len(inds)
```

```python
k=298
plt.plot(phi[inds[k]:inds[k+1]],H[inds[k]:inds[k+1]])
np.min(H[inds[k]:inds[k+1]])
#inds[k+1]-inds[k]
```

```python
from scipy import interpolate
temp=[]
for k in range(0,299):
    f = interpolate.interp1d(phi[inds[k]:inds[k+1]], H[inds[k]:inds[k+1]], kind='cubic')
    phinew=np.linspace(0.,2,100)*np.pi
    Hnew=f(phinew)
    temp.append(Hnew)
Hphi1=np.asarray(temp)
Hphi1.shape
np.savetxt('Hphi1.txt',Hphi1)
```

```python
phinew=np.linspace(0.,2,100)*np.pi
k=298
plt.plot(phinew,Hphi1[k,:])
```

```python
rr=np.loadtxt('rr.txt')
```

```python
f=interpolate.interp2d(rr,phinew,hphi2.T)#,fill_value="extrapolate")
```

```python
r1=np.linspace(1.1,50,200)
```

```python
surfH=f(r1,phinew)
```

```python
from mpl_toolkits import mplot3d
ax=plt.axes(projection='3d')
rr,pp=np.meshgrid(r1,phinew)
xx=rr*np.cos(pp)
yy=rr*np.sin(pp)
ax.plot_surface(xx,yy,surfH)
```
